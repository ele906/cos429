```python
import numpy as np
import cv2
from matplotlib import pyplot as plt
```

```python
img = cv2.imread('test.jpg', 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.xticks([]), plt.yticks([])
plt.show() 
```

```python
grayscale = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(cv2.cvtColor(grayscale, cv2.COLOR_GRAY2RGB))
plt.xticks([]), plt.yticks([])
plt.show() 
```

```python
height, width = img.shape[:2]
print(height, width)
middle = img[height//2-50:height//2+50, width//2-50:width//2+50, :]
plt.imshow(cv2.cvtColor(middle, cv2.COLOR_BGR2RGB))
plt.xticks([]), plt.yticks([])
plt.show() 
```

```python
zero = img.copy()

for i in range(0, width-1):
    for j in range(0, height-1):
        if (i % 10 == 0 and j % 10 == 0):
            zero[j, i, :] = 0

plt.imshow(cv2.cvtColor(zero, cv2.COLOR_BGR2RGB))
plt.xticks([]), plt.yticks([])
plt.show() 
```

```python
flip = cv2.flip(img, 1)

wide = np.zeros((height, width * 2, 3), dtype=np.uint8)

wide[:, width:] = img
wide[:, :width] = flip
plt.imshow(cv2.cvtColor(wide, cv2.COLOR_BGR2RGB))
plt.xticks([]), plt.yticks([])
plt.show() 

cv2.imwrite('output.jpg', wide)
```
